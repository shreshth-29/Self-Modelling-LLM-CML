```python
import os, re, json, random, unicodedata, urllib.request, zipfile
from datasets import load_dataset
os.chdir("/workspace/shreshth")
SEED = 42
GEO = re.compile(r"[\u10A0-\u10FF\u1C90-\u1CBF]")          # Georgian letters

def norm(s):
    return unicodedata.normalize("NFC", s or "").strip()

def opus_moses(corpus, version, src="en", tgt="ka"):
    """Download an OPUS bitext (two aligned text files in a zip)."""
    url = f"https://object.pouta.csc.fi/OPUS-{corpus}/{version}/moses/{src}-{tgt}.txt.zip"
    os.makedirs("data/raw", exist_ok=True)
    path = f"data/raw/{corpus}.{src}-{tgt}.zip"
    if not os.path.exists(path):
        urllib.request.urlretrieve(url, path)
    with zipfile.ZipFile(path) as z:
        names = z.namelist()
        en = z.read(next(n for n in names if n.endswith(f".{src}"))).decode("utf-8", "ignore").split("\n")
        ka = z.read(next(n for n in names if n.endswith(f".{tgt}"))).decode("utf-8", "ignore").split("\n")
    return list(zip(en, ka))

def clean(rows):
    """Same rules as before: <=250 chars, >=3 English words, target really Georgian,
    sane length ratio, no duplicates."""
    out, seen = [], set()
    for en, ka in rows:
        en, ka = norm(en), norm(ka)
        if not en or not ka or len(en) > 250 or len(ka) > 250 or len(en.split()) < 3:
            continue
        if not GEO.search(ka) or ka == en:
            continue
        r = len(ka) / len(en)
        if r < 0.5 or r > 2.5 or (en, ka) in seen:
            continue
        seen.add((en, ka))
        out.append({"en": en, "tgt": ka})
    return out

raw = {}
try:
    ds = load_dataset("Helsinki-NLP/opus-100", "en-ka", split="train")
    raw["OPUS-100"] = [(ex["translation"]["en"], ex["translation"]["ka"]) for ex in ds]
except Exception as e:
    print(f"OPUS-100 en-ka unavailable: {type(e).__name__}: {e}")
for corpus, version in [("TED2020", "v1"), ("WikiMatrix", "v1"), ("NLLB", "v1")]:
    try:
        raw[corpus] = opus_moses(corpus, version)
    except Exception as e:
        print(f"{corpus} unavailable: {type(e).__name__}: {e}")

cleaned = {k: clean(v) for k, v in raw.items()}
for k in raw:
    print(f"{k:12s}: {len(raw[k]):>10,} raw -> {len(cleaned[k]):>10,} after cleaning")
```
